```python
#Visualization of the distribution of the target variable
sns.countplot(biome_fecalcal, x="diagnosis", hue="fecalcal_gt_150")
plt.title("Count of Inflammation Score Greater than 150", fontsize=16)
```

```python
#Set y axis range for # Plot for Healthy Individuals (diagnosis = 1) to the same as Plot for Crohns and Colitis
crohns_fecalcal_gt_150 = biome_fecalcal[(biome_fecalcal['fecalcal'] > 150) & (biome_fecalcal['diagnosis'] == 'Crohns Disease')]
colitis_fecalcal_gt_150 = biome_fecalcal[(biome_fecalcal['fecalcal'] > 150) & (biome_fecalcal['diagnosis'] == 'Ulcerative Colitis')]
healthy_fecalcal_gt_150 = biome_fecalcal[(biome_fecalcal['fecalcal'] > 150) & (biome_fecalcal['diagnosis'] == 'Healthy')]

#Code by Gemini
#Determine a common set of bins and find the maximum count across all histograms
fecalcal_data_all_gt_150 = biome_fecalcal[biome_fecalcal['fecalcal'] > 150]['fecalcal']

max_y_lim = 10 # Default minimum y-limit
bins = 10 # Default number of bins

if not fecalcal_data_all_gt_150.empty:
    _, bins = np.histogram(fecalcal_data_all_gt_150, bins=bins) # Use a common binning strategy

    max_count = 0
    for subset_df in [crohns_fecalcal_gt_150, colitis_fecalcal_gt_150, healthy_fecalcal_gt_150]:
        if not subset_df.empty:
            counts, _ = np.histogram(subset_df['fecalcal'], bins=bins)
            max_count = max(max_count, np.max(counts))
    # Add a small buffer to the max_count for better visualization
    max_y_lim = max_count * 1.1 if max_count > 0 else 10

# Define a color palette for the diagnosis names
color_palette = {'Crohns Disease': 'red', 'Ulcerative Colitis': 'blue', 'Healthy': 'green'}

fig, axes = plt.subplots(1, 3, figsize=(20, 8))

# Plot for Crohns Disease
sns.histplot(crohns_fecalcal_gt_150, x="fecalcal", hue='diagnosis', multiple='dodge', ax=axes[0], bins=bins, palette=color_palette)
axes[0].set_title('Distribution of Fecal Cal > 150 in Crohns Disease', fontsize=14)
axes[0].set_xlabel('Fecal Cal Score')
axes[0].set_ylabel('Count')
axes[0].set_ylim(0, max_y_lim)

# Plot for Ulcerative Colitis
sns.histplot(colitis_fecalcal_gt_150, x="fecalcal", hue='diagnosis', multiple='dodge', ax=axes[1], bins=bins, palette=color_palette)
axes[1].set_title('Distribution of Fecal Cal > 150 in Ulcerative Colitis', fontsize=14)
axes[1].set_xlabel('Fecal Cal Score')
axes[1].set_ylabel('Count')
axes[1].set_ylim(0, max_y_lim)

# Plot for Healthy Individuals
sns.histplot(healthy_fecalcal_gt_150, x="fecalcal", hue='diagnosis', multiple='dodge', ax=axes[2], bins=bins, palette=color_palette)
axes[2].set_title('Distribution of Fecal Cal > 150 in Healthy Individuals', fontsize=14)
axes[2].set_xlabel('Fecal Cal Score')
axes[2].set_ylabel('Count')
axes[2].set_ylim(0, max_y_lim)

plt.tight_layout()
plt.show()
```

# **Data Preparation/Feature Engineering**

**Create Dataset with top 20 bacteria by mean and median in each diagnosis**


*   Dataset to predict Diagnones by Bacterial Relative Abundance



```python
crohns_top_mean_features = crohns_summary.sort_values(by='mean', ascending=False).head(20)['Features'].tolist()
crohns_top_max_features = crohns_summary.sort_values(by='max', ascending=False).head(20)['Features'].tolist()

colitis_top_mean_features = colitis_summary.sort_values(by='mean', ascending=False).head(20)['Features'].tolist()
colitis_top_max_features = colitis_summary.sort_values(by='max', ascending=False).head(20)['Features'].tolist()

healthy_top_mean_features = healthy_summary.sort_values(by='mean', ascending=False).head(20)['Features'].tolist()
healthy_top_max_features = healthy_summary.sort_values(by='max', ascending=False).head(20)['Features'].tolist()
```

```python
# Combine all unique top bacteria features
all_top_bacteria = list(set(crohns_top_mean_features + crohns_top_max_features +
                            colitis_top_mean_features + colitis_top_max_features +
                            healthy_top_mean_features + healthy_top_max_features))

# Add 'External ID' and 'diagnosis' back for identification if needed
# Ensure 'diagnosis' is in atlas_biome_bacteria before adding
features_to_include = ['External ID', 'diagnosis'] + all_top_bacteria

# Filter the atlas_biome_bacteria to create the new dataset
# Ensure all features_to_include actually exist in atlas_biome_bacteria.columns
existing_features_to_include = [f for f in features_to_include if f in atlas_biome.columns]

top_bacteria_df = atlas_biome[existing_features_to_include]

display(top_bacteria_df.head())
```

```python
top_bacteria_df.info()
```

```python
top_bacteria_df['diagnosis'] = top_bacteria_df['diagnosis'].astype('category').cat.codes
top_bacteria_df['External ID'] = top_bacteria_df['External ID'].astype('category').cat.codes
```
